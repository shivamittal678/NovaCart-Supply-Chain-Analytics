ABC Analysis

```python
abc=(
    inventory.groupby("SKU_ID")
    .agg(
        Sales_Qty=("Sales_Qty","sum"),
        Unit_cost=("Unit_Cost_INR","sum"),

    )
    .reset_index()
)
```

```python
abc
```

Consumption Value

```python
abc["Consumption_Value"]=(
    abc["Sales_Qty"]
    *
    abc['Unit_cost']
)
```

```python
abc=abc.sort_values(
    "Consumption_Value",
    ascending=False
)
```

```python
abc["cumulative_Pct"]=(
    abc["Consumption_Value"]
    .cumsum()
    /
    abc["Consumption_Value"].sum()
    *100
)
```

```python
abc["Consumption_Value"].dtype
```

```python
abc["Consumption_Value"] = pd.to_numeric(
    abc["Consumption_Value"],
    errors="coerce"
)
```

classify A,B and C


```python
def abc_class(x):
    if x<=80:
        return "A"
    elif x<=95:
        return "B"
    return "C"
```

```python
abc["ABC_Class"]=(
    abc['cumulative_Pct']
    .apply(abc_class)
)
```

```python
import pandas as pd

# 1. Take only the columns we need
temp = inventory[["SKU_ID", "Sales_Qty", "Unit_Cost_INR"]].copy()

# 2. Clean Sales_Qty
temp["Sales_Qty"] = pd.to_numeric(
    temp["Sales_Qty"],
    errors="coerce"
)

# 3. IMPORTANT: remove commas from Unit Cost first
temp["Unit_Cost_INR"] = (
    temp["Unit_Cost_INR"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .str.strip()
)

# 4. Convert Unit Cost to number
temp["Unit_Cost_INR"] = pd.to_numeric(
    temp["Unit_Cost_INR"],
    errors="coerce"
)

# 5. Create one row per SKU
abc = (
    temp.groupby("SKU_ID", as_index=False)
    .agg(
        Sales_Qty=("Sales_Qty", "sum"),
        Unit_Cost_INR=("Unit_Cost_INR", "first")
    )
)

# 6. Calculate Consumption Value
abc["Consumption_Value"] = (
    abc["Sales_Qty"] * abc["Unit_Cost_INR"]
)

# 7. Sort highest consumption first
abc = abc.sort_values(
    "Consumption_Value",
    ascending=False
).reset_index(drop=True)

# 8. Calculate cumulative %
abc["Cumulative_Pct"] = (
    abc["Consumption_Value"].cumsum()
    / abc["Consumption_Value"].sum()
    * 100
).round(2)

# 9. Create ABC classification
abc["ABC_Class"] = pd.cut(
    abc["Cumulative_Pct"],
    bins=[0, 80, 95, 100],
    labels=["A", "B", "C"],
    include_lowest=True
)

# 10. Check the result
print(abc)
```
